# FracVAL 3D visualization

This notebook uses the reusable functions in `plot_aggregate.py`. Run `make test` from the project root first if the example `.dat` files are not present.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if ROOT.name == 'plot':
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'plot'))

# from plot_aggregate import load_aggregate, plot_plotly
from fracval import load_data, plot_3d 
```

## Monodisperse example

```python
mono_file = ROOT / 'tests' / 'monodisperse' / 'results' / 'N_00000100_Agg_00000001.dat'
mono = load_data(mono_file)
print(f'{len(mono)} particles; radius range = {mono[:,3].min():.4g} to {mono[:,3].max():.4g}')
fig = plot_3d(mono, mode='spheres', sphere_resolution=12)
fig.show()
```

## Polydisperse example

```python
poly_file = ROOT / 'tests' / 'polydisperse' / 'results' / 'N_00000100_Agg_00000001.dat'
poly = load_data(poly_file)
print(f'{len(poly)} particles; radius range = {poly[:,3].min():.4g} to {poly[:,3].max():.4g}')
fig = plot_3d(poly, mode='spheres', sphere_resolution=12)
fig.show()
```

### Faster rendering for large aggregates

For very large `N`, replace `mode='spheres'` with `mode='centers'`. Center mode uses marker size to indicate radius and is much lighter in the browser.

```python
from fracval import FracVALConfig, generate, save_bundle, load_bundle

cfg = FracVALConfig(n=100, df=1.79, kf=1.40, seed=12345)
agg = generate(cfg)

paths = save_bundle(agg, "run_001", stem="aggregate_0001")
print(paths)

restored = load_bundle(paths["json"])
print(type(restored))
print(restored.data.shape, restored.radius_of_gyration)
```
